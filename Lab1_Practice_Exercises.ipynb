### B10. Exception Handling — Safer Billing Counter
Write `bill_and_collect_cash(cart, menu, cash_paid, is_parcel=False)` that: (a) raises/catches a `KeyError` if an item in the cart isn't in `menu`, (b) raises/catches a `ValueError` if `cash_paid` is less than the bill total, and (c) otherwise prints the change due. Test all three cases.

````python
# TODO: B10 solution
def bill_and_collect_cash(cart, menu, cash_paid, is_parcel=False):
    try:
        total = calculate_bill(cart, menu, is_parcel)

        if cash_paid < total:
            raise ValueError("Insufficient Cash")

        balance = cash_paid - total

        print("Total Bill:", total)
        print("Balance:", balance)

    except Exception as e:
        print("Error:", e)


bill_and_collect_cash(["Puri", "Tea"], menu, 200, True)
````

Bring B1–B10 together into one working canteen billing program that:

2. Lets you build a **cart (list)** for a customer.
3. Uses your **`calculate_bill()` function** (from B8) to bill the cart, including discount and packing charge logic.
4. **Saves** the receipt to `my_canteen_sales_log.txt` (from B9).
5. Wraps the whole transaction in **try/except** (from B10) so an invalid item or insufficient cash doesn't crash the program.
6. Bills **at least 3 different customers/carts** in a row, appending each to the same log file, then prints the full day's log at the end.


````python
# TODO: Part C solution

menu = {
    "Samosa": 15.0,
    "Vada": 12.0,
    "Tea": 10.0,
    "Cold Coffee": 30.0,
    "Dosa": 40.0,
    "Idli": 20.0
}


def calculate_bill(cart, menu, is_parcel=False, gst_rate=0.05):
    subtotal = 0

    for item in cart:
        subtotal += menu[item]

    if subtotal > 200:
        discount = subtotal * 0.10
    elif subtotal > 100:
        discount = subtotal * 0.05
    else:
        discount = 0

    gst = (subtotal - discount) * gst_rate
    packing_charge = 5 if is_parcel else 0
    total = subtotal - discount + gst + packing_charge

    return {
        "subtotal": subtotal,
        "discount": discount,
        "gst": gst,
        "packing_charge": packing_charge,
        "total": total
    }


def save_receipt(customer_name, bill):
    with open("my_canteen_sales_log.txt", "a") as file:
        file.write(f"{customer_name}: Rs.{bill['total']:.2f}\n")


orders = [
    ("Ravi", ["Samosa", "Tea", "Dosa"], False, 100),
    ("Ananya", ["Dosa", "Cold Coffee"], True, 100),
    ("Kiran", ["Vada", "Idli", "Tea"], False, 50)
]

for customer, cart, is_parcel, cash_paid in orders:
    try:
        bill = calculate_bill(cart, menu, is_parcel)

        if cash_paid < bill["total"]:
            raise ValueError("Insufficient cash")

        change = cash_paid - bill["total"]

        print("\nCustomer:", customer)
        print("Cart:", cart)
        print("Total:", bill["total"])
        print("Change:", change)

        save_receipt(customer, bill)

    except KeyError as e:
        print("Invalid item:", e)

    except ValueError as e:
        print("Error:", e)


print("\nFull Sales Log:")
with open("my_canteen_sales_log.txt", "r") as file:
    print(file.read())
````

## Stretch Challenge (Hard) — Canteen Rush Hour System

You've already handled Python before, so this one skips the warm-up: it's B1–B10 and Part C combined into a single messier, real-world-shaped problem. Nothing here uses any concept you haven't already touched today — the difficulty is entirely in **combining them correctly** and **handling the edge cases without letting one bad order crash the whole run**.

### Scenario
It's lunch rush. Multiple customers place orders in quick succession. Some orders will have mistakes (an item not on the menu, or not enough cash). Your program must process the whole rush, skip and log the bad orders, and still produce an accurate summary at the end.

### Setup
```python
menu = {
    "Samosa": {"price": 15.0, "category": "Snacks"},
    "Vada": {"price": 12.0, "category": "Snacks"},
    "Tea": {"price": 10.0, "category": "Beverages"},
    "Cold Coffee": {"price": 30.0, "category": "Beverages"},
    "Dosa": {"price": 40.0, "category": "Meals"},
    "Idli": {"price": 20.0, "category": "Meals"}
}

orders = [
    {"customer": "Ravi",   "cart": ["Samosa", "Tea"],       "tier": "regular", "hour": 13, "is_parcel": False, "cash_paid": 30},
    {"customer": "Ananya", "cart": ["Dosa", "Cold Coffee"], "tier": "gold",    "hour": 12, "is_parcel": True,  "cash_paid": 100},
    {"customer": "Kiran",  "cart": ["Vada", "Idli"],        "tier": "silver",  "hour": 14, "is_parcel": False, "cash_paid": 45},
    {"customer": "Meena",  "cart": ["Pizza"],               "tier": "regular", "hour": 13, "is_parcel": False, "cash_paid": 100},
    {"customer": "Arjun",  "cart": ["Dosa", "Dosa", "Tea"], "tier": "gold",    "hour": 12, "is_parcel": False, "cash_paid": 95},
    {"customer": "Divya",  "cart": ["Tea"],                 "tier": "regular", "hour": 15, "is_parcel": False, "cash_paid": 5},
]
```

### Requirements

Write `process_order(order, menu)` and a driver loop that, across all orders in `orders`:

1. **Peak-hour surcharge** — if `hour` is 12 or 13 (lunch rush), add an 8% surcharge on the subtotal *before* GST. Outside those hours, no surcharge.
2. **Two competing discounts, pick the better one** — a membership discount (`regular`: 0%, `silver`: 5%, `gold`: 12%) *and* the loyalty discount from earlier (subtotal > 200 → 10%, > 100 → 5%, else 0%). Apply whichever discount is **larger**, never both stacked.
3. **GST (5%)** applies after the discount, and a **Rs.5 packing charge** if `is_parcel` is True — same as before.
4. **Bad orders must not stop the run.** If a cart item isn't in `menu`, catch it (`KeyError`), log the order to `failed_transactions.txt` with the reason, and move on to the next order. Same for insufficient cash (`ValueError`) — don't let the loop die.
5. **Successful orders** get appended to `rush_hour_log.txt` as `"<customer>: Rs.<total>"`.
6. **End-of-rush summary**, printed (not just written to file):
   - Total revenue collected across all successful orders.
   - Revenue broken down **by category** (Snacks / Beverages / Meals) — build this without importing anything, using a plain dictionary you update as you go.
   - The single **best-selling item** by total quantity sold across all orders (again, no imports — track counts in a dictionary).
   - How many orders **failed**, and why.

Try to get through all 5 sample orders (including the two broken ones) without your program stopping partway.


````python
# TODO: Stretch challenge solution
menu = {
    "Samosa": {"price": 15.0, "category": "Snacks"},
    "Vada": {"price": 12.0, "category": "Snacks"},
    "Tea": {"price": 10.0, "category": "Beverages"},
    "Cold Coffee": {"price": 30.0, "category": "Beverages"},
    "Dosa": {"price": 40.0, "category": "Meals"},
    "Idli": {"price": 20.0, "category": "Meals"}
}

orders = [
    {"customer": "Ravi", "cart": ["Samosa", "Tea"], "tier": "regular", "hour": 13, "is_parcel": False, "cash_paid": 30},
    {"customer": "Ananya", "cart": ["Dosa", "Cold Coffee"], "tier": "gold", "hour": 12, "is_parcel": True, "cash_paid": 100},
    {"customer": "Kiran", "cart": ["Vada", "Idli"], "tier": "silver", "hour": 14, "is_parcel": False, "cash_paid": 45},
    {"customer": "Meena", "cart": ["Pizza"], "tier": "regular", "hour": 13, "is_parcel": False, "cash_paid": 100},
    {"customer": "Arjun", "cart": ["Dosa", "Dosa", "Tea"], "tier": "gold", "hour": 12, "is_parcel": False, "cash_paid": 95},
    {"customer": "Divya", "cart": ["Tea"], "tier": "regular", "hour": 15, "is_parcel": False, "cash_paid": 5}
]


def process_order(order, menu):
    subtotal = 0
    category_sales = {}
    item_counts = {}

    for item in order["cart"]:
        if item not in menu:
            raise KeyError(item)

        price = menu[item]["price"]
        category = menu[item]["category"]

        subtotal += price

        if category not in category_sales:
            category_sales[category] = 0
        category_sales[category] += price

        if item not in item_counts:
            item_counts[item] = 0
        item_counts[item] += 1

    membership_discount_rates = {
        "regular": 0,
        "silver": 0.05,
        "gold": 0.12
    }

    membership_discount = (
        subtotal * membership_discount_rates[order["tier"]]
    )

    if subtotal > 200:
        loyalty_discount = subtotal * 0.10
    elif subtotal > 100:
        loyalty_discount = subtotal * 0.05
    else:
        loyalty_discount = 0

    discount = max(membership_discount, loyalty_discount)

    discounted_subtotal = subtotal - discount

    if order["hour"] == 12 or order["hour"] == 13:
        surcharge = subtotal * 0.08
    else:
        surcharge = 0

    gst = (discounted_subtotal + surcharge) * 0.05

    packing_charge = 5 if order["is_parcel"] else 0

    total = discounted_subtotal + surcharge + gst + packing_charge

    if order["cash_paid"] < total:
        raise ValueError("Insufficient cash")

    change = order["cash_paid"] - total

    return {
        "subtotal": subtotal,
        "discount": discount,
        "surcharge": surcharge,
        "gst": gst,
        "packing_charge": packing_charge,
        "total": total,
        "change": change,
        "category_sales": category_sales,
        "item_counts": item_counts
    }


total_revenue = 0
category_revenue = {
    "Snacks": 0,
    "Beverages": 0,
    "Meals": 0
}
item_counts = {}
failed_orders = []

open("rush_hour_log.txt", "w").close()
open("failed_transactions.txt", "w").close()

for order in orders:
    try:
        bill = process_order(order, menu)

        total_revenue += bill["total"]

        for category in bill["category_sales"]:
            category_revenue[category] += bill["category_sales"][category]

        for item in bill["item_counts"]:
            if item not in item_counts:
                item_counts[item] = 0
            item_counts[item] += bill["item_counts"][item]

        with open("rush_hour_log.txt", "a") as file:
            file.write(
                f"{order['customer']}: Rs.{bill['total']:.2f}\n"
            )

        print(
            order["customer"],
            "successful - Total:",
            round(bill["total"], 2),
            "Change:",
            round(bill["change"], 2)
        )

    except KeyError as e:
        reason = f"Item not found: {e}"
        failed_orders.append((order["customer"], reason))

        with open("failed_transactions.txt", "a") as file:
            file.write(f"{order['customer']}: {reason}\n")

        print(order["customer"], "failed:", reason)

    except ValueError as e:
        reason = str(e)
        failed_orders.append((order["customer"], reason))

        with open("failed_transactions.txt", "a") as file:
            file.write(f"{order['customer']}: {reason}\n")

        print(order["customer"], "failed:", reason)


best_item = None
best_quantity = 0

for item in item_counts:
    if item_counts[item] > best_quantity:
        best_item = item
        best_quantity = item_counts[item]


print("\n----- END OF RUSH SUMMARY -----")
print("Total Revenue:", round(total_revenue, 2))

print("\nRevenue by Category:")
for category in category_revenue:
    print(category, ":", round(category_revenue[category], 2))

print("\nBest-Selling Item:")
print(best_item, "-", best_quantity, "sold")

print("\nFailed Orders:", len(failed_orders))

for customer, reason in failed_orders:
    print(customer, "-", reason)
````
