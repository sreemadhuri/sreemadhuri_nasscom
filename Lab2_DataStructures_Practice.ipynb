### 1. Lists — Lab Submissions Log
Create a list `lab_submissions` with at least 8 entries (with some repeats — e.g. `"OS"` appearing 3 times, representing multiple lab submissions for that subject). Then:

### 2. Tuples — Fixed Exam Slot
Create a tuple `exam_slot = ("Computer Networks", "2:00 PM")`. Unpack it into `subject` and `time` variables and print them. Then try to change the time (`exam_slot[1] = "3:00 PM"`) inside a `try/except` block and print the error message you get — this proves tuples are immutable.

```python
# TODO: 2. Tuples solution
exam_slot = ("Computer Networks", "2:00 PM")

subject, time = exam_slot

print("Subject:", subject)
print("Time:", time)

try:
    exam_slot[1] = "3:00 PM"
except TypeError as e:
    print("Error:", e)
```

### 5. Sets — Comparing Two Students' Skills
Create two sets, `your_skills` and `friend_skills`, each with at least 4 programming languages/tools (with some overlap). Print:
- Skills you **both** know (intersection)
- All skills **across both of you** (union)
- Skills only **you** know (difference)

```python
# TODO: 5. Sets solution
your_skills = {"Python", "C", "Java", "HTML", "Git"}
friend_skills = {"Python", "Java", "JavaScript", "Git", "SQL"}

print("Skills both know:", your_skills.intersection(friend_skills))
print("All skills:", your_skills.union(friend_skills))
print("Skills only you know:", your_skills.difference(friend_skills))
```

### 6. Dictionaries — Nested Gradebook
Create a nested dictionary `gradebook` where each subject maps to a dictionary with `marks`, `credits`, and `faculty`. Include at least 5 subjects. Then:
- Print one subject's marks using double-key lookup (`gradebook["Subject"]["marks"]`).
- Use `.get()` to safely look up a subject that **isn't** in the gradebook, with a fallback message.

```python
# TODO: 6. Dictionaries solution
gradebook = {
    "Python": {"marks": 92, "credits": 4, "faculty": "Anita"},
    "DS": {"marks": 88, "credits": 4, "faculty": "Ravi"},
    "DBMS": {"marks": 81, "credits": 3, "faculty": "Priya"},
    "Java": {"marks": 90, "credits": 4, "faculty": "Kiran"},
    "OS": {"marks": 76, "credits": 3, "faculty": "Meena"}
}

print("Python marks:", gradebook["Python"]["marks"])

print(
    gradebook.get(
        "AI",
        "Subject not found in gradebook"
    )
)
```

### 7. Dictionary Comprehension
Using the `gradebook` from Exercise 6, build a simple `{subject: marks}` dictionary in **one line** using a dictionary comprehension.

```python
# TODO: 7. Dictionary comprehension solution
subject_marks = {subject: details["marks"] for subject, details in gradebook.items()}

print(subject_marks)
```

### 8. Set Comprehension — High-Scoring Subjects
Using your `gradebook` (Exercise 6) and the unique subjects submitted (Exercise 4), build a **set comprehension** that keeps only the subjects submitted this week where marks are 85 or above.

```python
# TODO: 8. Set comprehension solution
high_scoring = {
    subject
    for subject in unique_subjects
    if subject in gradebook and gradebook[subject]["marks"] >= 85
}

print("High-scoring subjects:", high_scoring)
```

## Combined Task — Semester Report (submit this)

Bring everything together into one program that:
1. Defines a nested `gradebook` **dictionary** (at least 6 subjects, with `marks` and `credits`).
2. Defines a `lab_submissions` **list** with at least 10 entries (with repeats).
3. Computes the **unique subjects submitted** using a **set**.
4. Computes the **high-scoring subjects (marks >= 85)** using a **set comprehension**.
5. Computes the **best subject** as a **tuple** `(subject_name, marks)` — the highest-scoring subject overall.
6. Prints all of the above in a clearly labelled semester report.


```python
# TODO: Combined task solution
gradebook = {
    "Python": {"marks": 92, "credits": 4},
    "DS": {"marks": 88, "credits": 4},
    "DBMS": {"marks": 81, "credits": 3},
    "Java": {"marks": 90, "credits": 4},
    "OS": {"marks": 76, "credits": 3},
    "Maths": {"marks": 87, "credits": 4}
}

lab_submissions = [
    "Python", "DS", "OS", "Python", "Java",
    "DBMS", "Maths", "OS", "DS", "Python"
]

unique_subjects = set(lab_submissions)

high_scoring = {
    subject
    for subject in unique_subjects
    if gradebook[subject]["marks"] >= 85
}

best_subject = max(
    gradebook.items(),
    key=lambda x: x[1]["marks"]
)

best_subject = (best_subject[0], best_subject[1]["marks"])

print("===== SEMESTER REPORT =====")
print("Gradebook:", gradebook)
print("Lab Submissions:", lab_submissions)
print("Unique Subjects:", unique_subjects)
print("High-Scoring Subjects:", high_scoring)
print("Best Subject:", best_subject)


```
